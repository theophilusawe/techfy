```python
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Load the data

```python
from sklearn.datasets import make_regression

X, y = make_regression(n_samples=1000, n_features=1, n_targets=1, noise=30, random_state=42)
```

## 2. Data Preparation

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

## 3. Model Training

Instantiate the model

Fit the model to our data

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(X_train, y_train)

print("Model trained successfully")
```

## 4. Model Prediction

```python
predictions = model.predict(X_test)
```

```python
predictions[:5], y_test[:5]
```

## 5. Model Evaluation

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np

mae = mean_absolute_error(y_test, predictions)
print(f' The Mean Absolute Error is {mae:.3f}\n')

mse = mean_squared_error(y_test, predictions)
print(f' The Mean Squared Error is {mse:.3f}\n')

rscore = r2_score(y_test, predictions)
print(f' The r2_score is {rscore:.3f}\n')

rmse = np.sqrt(mse)
print(f' The Root Mean Squared Error is {rmse:.3f}\n')
```

##  6. Visualization

```python
plt.figure(figsize=(10,6))
plt.scatter(X_test, y_test, color='blue', alpha=.4)
plt.plot(X_test, predictions, 'r--', lw=2)
plt.show()
```
