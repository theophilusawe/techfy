```python
import matplotlib.pyplot as plt
```

```python
fig, axes = plt.subplots(3, 2, figsize=(10,6))
fig.suptitle('Quarterly Financial Performance', fontsize=16, fontweight='bold')
# fig.tight_layout(pad=4.0)
fig.subplots_adjust(wspace=0.4, hspace=0.6)
axes[0][0].plot([.2,.3],[.2,.4], 'r--')
axes[0][1].plot([.2,.4],[.1,.2], 'b^^')
fig.legend(labels=['Target', 'Actual'], loc='lower center')

for i, ax in enumerate(axes):
    axa, axb = ax
    axa.set_title(f'Quarter {i}')
    axb.set_title(f'Quarter {i+1}')

```

```python
fig.savefig('test2.png', dpi=300, bbox_inches='tight')
```
