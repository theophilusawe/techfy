```python
def strip_whitespace(df: pd.DataFrame) -> pd.DataFrame:
    for col in df.select_dtypes(include='object').columns:
        df[col] = df[col].str.strip()
    return df

census_df = strip_whitespace(census_df)
```

```python
def check_null_count(df: pd.DataFrame):
    result = {}
    for column in df.columns:
        null_count = df[column].isna().sum()
        question_mark_count = (df[column] == ' ?').sum()
        effective_null_count = null_count + question_mark_count
        result[column] = {
            "null_count" : int(null_count),
            'question_mark_count': int(question_mark_count),
            'effective_null_count': int(effective_null_count),
            '-------------------': '------------------------------'
        }
        

    return result
```
