```python
from sklearn.metrics import roc_auc_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, precision_score

data = load_breast_cancer()
X_train, X_test, y_train, y_test = train_test_split(
    data.data, data.target, test_size=0.2, random_state=42, stratify=data.target
)

knn_pipe = Pipeline([
    ('scaler', StandardScaler()),         # MUST scale for KNN
    ('model',  KNeighborsClassifier(n_neighbors=5, weights='distance'))
])

knn_pipe.fit(X_train, y_train)
print(classification_report(y_test, knn_pipe.predict(X_test),
      target_names=['Malignant', 'Benign']))

# Try different values of k and compare accuracy
for k in [1, 3, 5, 11, 21]:
    pipe = Pipeline([('s', StandardScaler()), ('m', KNeighborsClassifier(n_neighbors=k))])
    pipe.fit(X_train, y_train)
    y_pred = pipe.predict(X_test) 
    print(f"For {k} nearest neighbors, the roc_auc score is {roc_auc_score(y_pred, y_test)}")
    
```
